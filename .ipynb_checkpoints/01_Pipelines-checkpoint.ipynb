# pipeline() in Hugging Face
pipeline() makes it easy to run a model with all preprocessing and postprocessing steps.

You just give an input (like text) and get a result (like sentiment or summary).

Behind the scenes:

- Input is preprocessed

- Model gives predictions

- Output is post-processed

# Types of NLP Pipelines
## Text pipelines
text-generation: Generate text from a prompt \
text-classification: Classify text into predefined categories \
summarization: Create a shorter version of a text while preserving key information \
translation: Translate text from one language to another \
zero-shot-classification: Classify text without prior training on specific labels \
feature-extraction: Extract vector representations of text 
## Image pipelines
image-to-text: Generate text descriptions of images \
image-classification: Identify objects in an image \
object-detection: Locate and identify objects in images
## Audio pipelines
automatic-speech-recognition: Convert speech to text \
audio-classification: Classify audio into categories \
text-to-speech: Convert text to spoken audio 
## Multimodal pipelines
image-text-to-text: Respond to an image based on a text prompt

```python
from transformers import pipeline
```

# Zero-shot classification : No training needed; just give labels.

```python
classifier = pipeline("zero-shot-classification")
```

```python
classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

```python
classifier(
    "The battery life of this phone is terrible. It dies within 4 hours of use.",
    candidate_labels = ["positive", "negative", "neutral"]
)
```

```python
classifier(
    "NASA announces successful test of next-generation rocket engines.",
    candidate_labels = ["science", "politics", "entertainment", "sports"]
)
```

```python
classifier(
    "I need to talk to someone about my recent bill. I think I was overcharged.",
    candidate_labels = ["billing issue", "technical support", "general inquiry", "cancellation request"]
)
```

# Text generation: Auto-completes a sentence or prompt.

```python
generator = pipeline("text-generation")
```

```python
generator("In this course, we will teach you how to", max_length=30, num_return_sequences=2)
```

# Mask filling: Predicts missing word in a sentence.

```python
unmasker = pipeline("fill-mask")
```

```python
unmasker("This course will teach you all about <mask> models.", top_k=2)
```

# Named Entity Recognition (NER): Finds people, places, orgs in text.

```python
ner = pipeline("ner", grouped_entities=True)
```

```python
ner("My name is Sylvain and I work at Hugging Face in Brooklyn.")
```

# Question Answering: Finds answer from a given paragraph.

```python
question_answerer = pipeline("question-answering")
```

```python
question_answerer(
    question="Where do I work?",
    context="My name is Sylvain and I work at Hugging Face in Brooklyn",
)
```

```python
question_answerer = pipeline("question-answering", model="deepset/roberta-base-squad2")
```

```python
question_answerer(
    question="Where do I work?",
    context="My name is Sylvain and I work at Hugging Face in Brooklyn",
)
```

# Summarization: Makes text shorter, keeping important info.

```python
summarizer = pipeline("summarization")
```

```python
summarizer(
    """
    America has changed dramatically during recent years. Not only has the number of 
    graduates in traditional engineering disciplines such as mechanical, civil, 
    electrical, chemical, and aeronautical engineering declined, but in most of 
    the premier American universities engineering curricula now concentrate on 
    and encourage largely the study of engineering science. As a result, there 
    are declining offerings in engineering subjects dealing with infrastructure, 
    the environment, and related issues, and greater concentration on high 
    technology subjects, largely supporting increasingly complex scientific 
    developments. While the latter is important, it should not be at the expense 
    of more traditional engineering.

    Rapidly developing economies such as China and India, as well as other 
    industrial countries in Europe and Asia, continue to encourage and advance 
    the teaching of engineering. Both China and India, respectively, graduate 
    six and eight times as many traditional engineers as does the United States. 
    Other industrial countries at minimum maintain their output, while America 
    suffers an increasingly serious decline in the number of engineering graduates 
    and a lack of well-educated engineers.
"""
)
```

# Translation: Converts text between languages.

```python
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-fr-en")
```

```python
translator("Ce cours est produit par Hugging Face.")
```

# Image and audio pipelines

```python
image_classifier = pipeline(
    task="image-classification", model="google/vit-base-patch16-224"
)
```

```python
result = image_classifier(
    "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/pipeline-cat-chonk.jpeg"
)
print(result)
```

```python
transcriber = pipeline(
    task="automatic-speech-recognition", model="openai/whisper-large-v3"
)
```

```python
result = transcriber(
    "https://huggingface.co/datasets/Narsil/asr_dummy/resolve/main/mlk.flac"
)
print(result)
```
